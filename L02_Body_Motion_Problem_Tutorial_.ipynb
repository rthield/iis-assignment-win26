### Multi-Question Student Task Sheet (0.25 Pt x 6)


1. **E1 · The Gap & Whole-Course Judgment**
   - Eva says "Serve me a drink." A video shows a robot executing a smooth trajectory, placing the mug on the table, but a photograph reveals milk spilled across the counter and an open fridge left behind. Explain why this nominal end-state success does **not** constitute a solution to the Body Motion Problem, and specify two whole-course constraints that were violated.


**Ans:**
The end state (mug on table) looks right, but the Body Motion Problem asks that the **whole course** of the action achieve the goal while respecting constraints. The spill and the open fridge show that the route there failed the task.
- **Violated constraint 1, no spillage:** the milk must stay contained during grasp, carry and pour.
- **Violated constraint 2, restore the environment:** the carton must be put away and the fridge closed before the task is done.


2. **E2 · Constrained Action Designator & Grounding**

    - Distinguish interpretation, inferred means, and grounded objects for this episode by filling out the three missing slots in the schematic action designator:

- **Working Interpretation:** `(:type serving) (:recipient Eva) (:substance coffee-with-milk)`
- **Inferred Function/Means:** `(:vessel-role drink-container)`
- **Grounded Physical Object:** `(:object eva-mug)`


*Interpretation* is what Eva wants (she confirmed coffee with milk). *Inferred means* follows from reasoning: carrying a drink requires containment and support, so a vessel role is created. *Grounding* links that role to a perceived object: Eva's mug on the counter, which already holds black coffee.


3. **E3 · The Commitment Decision & Asymmetric Failure Costs**

    - During the retrieval step, the robot faces a choice between a **5 mm test lift near the shelf** or a **high carry directly to the counter**. State the predicted effect, the cost of failure, and the single piece of evidence that justifies choosing the low test lift before committing to the carry.


**Ans:**
- **5 mm test lift:** predicted effect is that the carton rises slightly and stays in the grip. Failure cost is low: it drops a few mm onto the shelf, nothing spills, and the robot can regrasp.
- **High carry:** failure cost is high and irreversible: the carton falls from height, milk spills and needs cleaning.
- **Justifying evidence:** the **wet surface on Carton A** makes grip friction uncertain, so a cheap test comes before the costly commitment.


4. **E4 · Causal Prediction across Material Variants**

    - Compare pouring milk into coffee versus pancake batter into a pan. Formulate the stopping rule for each liquid using `(:stop-when ...)`, and explain how material dynamics alter when the robot must command the wrist to tilt back.


**Ans:**
- **Milk:** `(:stop-when (>= (+ poured-volume in-flight-volume) 30ml))`. Milk is thin and flows fast. Milk still in the stream lands after the untilt, so the robot tilts back **slightly before** the target.
- **Batter:** `(:stop-when (>= predicted-final-diameter target-diameter))`. Batter is viscous, keeps dripping off the lip, and **keeps spreading** on the hot pan, so the robot tilts back **well before** the visible disc reaches the target size.

In both cases the stop is based on the predicted *settled* result, not a fixed angle or duration.


5. **E5 · System Decomposition & The Reasoning Loop**

    - Trace the carton slip event through the seven-stage reasoning loop: **Infer → Ground → Predict → Decide → Regulate → Detect & Revise → Verify**. State precisely which step was triggered when downward motion was observed during the 5 mm lift, and how the action designator was developed in response.


**Ans:**
- **Infer:** Eva wants coffee with milk, so milk must be fetched and poured.
- **Ground:** Carton A is found in the fridge door, and its wet surface is seen.
- **Predict:** a side pinch might slip, and a low lift will reveal this cheaply.
- **Decide:** do a 5 mm test lift.
- **Regulate:** lift while monitoring carton motion in the grip.
- **Detect & Revise:** ⟵ **triggered by the observed downward motion.**
- **Verify:** re-test the new grasp, then check the poured amount, carton returned, fridge closed and handover.

**Designator development:** the grasp changes from `(:grasp side-pinch)` to `(:grasp bottom-support)`, and a later `(:regrasp pourable-grasp)` step is added, because bottom support blocks the wrist tilt needed to pour. The carton is set down, regrasped and lifted again.


6. **E6 · Conceptual Ablation & Evidence of Competence**

    - Perform an ablation experiment. Suppose we remove the **"Check before commitment"** capability and replace it with a fixed default high-side-pinch lift. Predict a specific kitchen environment variant where this ablated system fails, and explain why single-run execution success differs from evidence of general competence across task variants.


**Ans:**
- **Failing variant:** a cold carton covered in condensation, full and heavy (like Carton A). The fixed high side pinch slips at height, the carton falls and milk spills. With no test lift, the robot has no cheap moment to notice. (On a dry carton both versions succeed, so they tie.)
- **Single run vs competence:** one success shows only that it worked in *that* world state, possibly by luck. Competence is shown by checked successes and failures across a range of variants (wet/dry, full/half, opened/unopened cartons).
